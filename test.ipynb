```python
# 1. MONTARE GOOGLE DRIVE (Opzionale)
# Rimuovi il '#' dalle due righe sotto se vuoi accedere ai tuoi file su Drive
# from google.colab import drive
# drive.mount('/content/drive')

# 2. IMPORTAZIONE DELLE LIBRERIE PRINCIPALI
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurazione per mostrare i grafici direttamente nel notebook
%matplotlib inline

# ==========================================
# 3. CARICAMENTO DATI (Esempio)
# ==========================================
# Creiamo un dataset di esempio (puoi sostituirlo con il tuo file CSV/Excel)
data = {
    'Età': [25, 30, 35, 40, 45, 50, 55],
    'Stipendio': [22000, 25000, 32000, 42000, 50000, 58000, 65000]
}
df = pd.DataFrame(data)

# Mostra le prime righe del dataset
print("--- Prime righe del DataFrame ---")
print(df.head())

# Mostra informazioni sul tipo di dati
print("\n--- Informazioni sul Dataset ---")
print(df.info())

# ==========================================
# 4. CREAZIONE DI UN GRAFICO DI ESEMPIO
# ==========================================
plt.figure(figsize=(8, 5))
sns.lineplot(x='Età', y='Stipendio', data=df, marker='o', color='blue')
plt.title('Andamento dello Stipendio in base all\'Età')
plt.xlabel('Età')
plt.ylabel('Stipendio (€)')
plt.grid(True)
plt.show()
```
